```python
import re

def clean_text(text):
    if pd.isna(text):
        return ""

    text = text.lower()

    # Replace symbols
    text = text.replace("&", " and ")

    # Remove common legal suffixes
    text = re.sub(r"\b(private|pvt|limited|ltd|corporation|corp|inc|llc|co)\b", " ", text)

    # Remove punctuation
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```
