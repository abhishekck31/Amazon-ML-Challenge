# Phase 1 - Exploratory Data Analysis

Loads every train/test source via `src.data_loader` and runs the reusable checks in `src.eda`: dataset summary, missing values, duplicates, country distribution, and sample records. No modeling here.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd

from src.data_loader import load_train, load_test
from src.eda import (
    dataset_summary,
    missing_value_report,
    duplicate_report,
    country_distribution,
    sample_records,
    generate_eda_report,
)
```

## Load data

Full files are ~1.9 GB combined. If you're short on free RAM, pass e.g. `nrows=200_000` to `load_train()`/`load_test()` instead of `None` while iterating.

```python
train = load_train(nrows=None)
test = load_test(nrows=None)
datasets = {**train, **test}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

## Dataset summary

```python
dataset_summary(datasets)
```

## Missing value report

```python
missing_value_report(datasets)
```

## Duplicate report

Full-row duplicates. `entity_id` is a synthetic key so it's expected to already be unique; this checks for accidental exact-duplicate records.

```python
duplicate_report(datasets)
```

## Country distribution

```python
country_distribution(datasets, top_n=15)
```

## Sample records from every source

```python
samples = sample_records(datasets, n=5)
for name, sample_df in samples.items():
    print(f"--- {name} ---")
    display(sample_df)
```

## Generate the written EDA report

Assembles all of the above into `output/eda_report.md`.

```python
report = generate_eda_report(datasets, output_path="../output/eda_report.md")
print("Report written to ../output/eda_report.md")
```
